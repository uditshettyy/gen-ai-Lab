```python
import nltk

from nltk.corpus import gutenberg, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('gutenberg')              #dataset

nltk.download('punkt')                  #for tokenization
nltk.download('punkt_tab')

nltk.download('stopwords')              #to remove stop words
nltk.download('wordnet')               #for vectorization
nltk.download('omw-1.4')
```

```python
print("Available Gutenberg Files : ")
print(gutenberg.fileids())
```

Selecting few docs from the Gutenberg dataset

```python
documents = [
    gutenberg.raw('austen-emma.txt'),
    gutenberg.raw('austen-persuasion.txt'),
    gutenberg.raw('austen-sense.txt')
]

print("Number of documents : ", len(documents))
```

Text preprocessing

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess_words(text):

    #Convert to lowercase (Normalization)
    text = text.lower()

    #Tokenization
    tokens  = word_tokenize(text)

    #Removing punctuations
    tokens = [word for word in tokens if word.isalpha()]

    #Removing stopwords (most common words such as : the, a, etc..)
    tokens = [word for word in tokens if word not in stop_words]

    #Lemmatization (reducing the length of the words)
    tokens = [lemmatizer.lemmatize(word) for word in tokens ]

    return " ".join(tokens)
```

Apply preprocessing

```python
cleaned_documents = []

for document in documents:
    cleaned_word = preprocess_words(document)
    cleaned_documents.append(cleaned_word)
```

Display before and after preprocessing

```python
print("Document before preprocessing : ")
print(documents[0][:1000])

print("\n")
print("=" *100)
print("\nDocument after preprocessing : ")
print(cleaned_documents[0][:1000])
```

Spliting the data into train and test split

```python
from sklearn.model_selection import train_test_split

x_train, x_test = train_test_split(cleaned_documents, test_size=0.25, random_state=0)
```

```python
print("Training documents : ", len(x_train))
print("Testing documents : ", len(x_test))
```

Create Bag of Words (words will be replaced with its frequency (like a hashmap :-))

```python
from sklearn.feature_extraction.text import CountVectorizer

bow_vectorizer = CountVectorizer()

#Fit only for train
x_train_bow = bow_vectorizer.fit_transform(x_train)

#Transform the test data using the same vocabulary
x_test_bow = bow_vectorizer.transform(x_test)
```

Displaying BOW (Bag Of Words) result

```python
print("Vocabulary size : ", len(bow_vectorizer.vocabulary_))

print("\nVocabulary : ")
print(bow_vectorizer.get_feature_names_out()[:50])

print("\nTraining matrix shape")
print(x_train_bow.shape)
```

```python
print("BOW Matrix : ")
print(x_train_bow.toarray()[:, :20])
```

TF - ID

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfid_vectorizer = TfidfVectorizer()

x_train_tfid = tfid_vectorizer.fit_transform(x_train)

x_test_tfid = tfid_vectorizer.transform(x_test)
```

```python
print("Vocabulary size : ", len(tfid_vectorizer.vocabulary_))

print("\nVocabulary : ")
print(tfid_vectorizer.get_feature_names_out()[:50])

print("\nTraining matrix shape")
print(x_train_tfid.shape)
```

```python
print("TF-IDF Matrix : ")
print(x_train_tfid.toarray()[:, :20])
```

Display selected words and their values

```python
feature_name = tfid_vectorizer.get_feature_names_out()

print("First 20 TF-IDF Features : ")
for i, word in enumerate(feature_name[:20]):
    print(f"{i} -> {word}")
```
