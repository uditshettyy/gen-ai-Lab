```python
import nltk 
from nltk.corpus import gutenberg,stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
nltk.download('gutenberg')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('own-1.4')
```

```python
print("Available Gutenberg Files:")
print(gutenberg.fileids())
```

```python
documents=[
    gutenberg.raw('austen-emma.txt'),
    gutenberg.raw('austen-persuasion.txt'),
    gutenberg.raw('austen-sense.txt'),

]
print("Number of Docs:",len(documents))
```

Text Preprosessing Function

```python
stop_words=set(stopwords.words('english'))
lemmatizer=WordNetLemmatizer()
def preprocess_text(text):
    # convet text to lowercase
    text=text.lower()
    # Tokenization 
    tokens=word_tokenize(text)
    # Remove punctuations
    tokens=[word for word in tokens if word.isalpha()]
    # Remove teh stop words 
    tokens=[word for word in tokens if word not in stop_words]
    # Lemmatization
    tokens=[lemmatizer.lemmatize(word) for word in tokens]

    return " ".join(tokens)
```

Apply Preprocessing

```python
cleaned_documents=[]

for documents in documents:
    cleaned_txt=preprocess_text(documents)
    cleaned_documents.append(cleaned_txt)
    
```

Display before and after prepocessing

```python
print("Before preprocessing:")
print(documents[0][:1000])
print("=============================")
print("After preprocesssing")
print(cleaned_documents[0][:1000])
```
